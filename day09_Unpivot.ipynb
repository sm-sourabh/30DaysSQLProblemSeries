```python
sales_df = spark.read.format("csv")\
    .option("Header","true")\
    .option("inferSchema","true")\
    .load("/Volumes/sql_problems/default/my_volume/day09_sales.csv")

display(sales_df)
```

```python
sales_df.createOrReplaceTempView("sales")
```

```python
%sql
DESCRIBE TABLE sales;
```

```python
%sql
SELECT 
    store_id,
    category,
    quarter,
    sales_amount
FROM sales
UNPIVOT (
    sales_amount
    FOR quarter IN (q1, q2, q3)
);
```

```python
from pyspark.sql.functions import coalesce, lit

unpivoted_df = (
    sales_df
    .unpivot(
        ["store_id", "category"],           #ids
        ["Q1", "Q2", "Q3"],                 #values
        "quarter",                          #variable_column
        "sales"                             #valueColumn
    )
    .withColumn("sales", coalesce(col("sales"), lit(0)))
)

display(unpivoted_df)
```
