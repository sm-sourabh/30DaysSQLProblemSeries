Task : Identify and extract all customers who have registered in the system (customers) but have never placed any orders (orders).

```python
customers_df = spark.read.format("csv")\
    .option("Header","true")\
    .option("inferSchema","true")\
    .load("/Volumes/sql_problems/default/my_volume/day10_customer.csv")

display(customers_df)
```

```python
orders_df = spark.read.format("csv")\
    .option("Header","true")\
    .option("inferSchema","true")\
    .load("/Volumes/sql_problems/default/my_volume/day10_order.csv")

display(orders_df)
```

```python

customers_df.createOrReplaceTempView("customers")
orders_df.createOrReplaceTempView("orders")
```

```python
%sql
SELECT c.customer_id, customer_name
FROM customers c
LEFT JOIN orders o 
ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL
```

```python
customers_df.printSchema()
orders_df.printSchema()
```

```python
from pyspark.sql.functions import *

joined_df = (
    customers_df.alias("c")
    .join(
        orders_df.alias("o"),
        col("c.customer_id") == col("o.customer_id"),
        "left_anti"
    )
)

display(joined_df)
```
