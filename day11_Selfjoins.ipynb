Task: Map each employee to their respective manager's name using a self-join to reveal organizational hierarchy and management structures.

```python
employee_df = spark.read.format("csv")\
    .option("Header","true")\
    .option("inferSchema","true")\
    .load("/Volumes/sql_problems/default/my_volume/day11_employee.csv")

display(employee_df)
```

```python
employee_df.createOrReplaceTempView("employees")
```

```python
%sql
SELECT 
    e.emp_id as EMP_id,
    e.emp_name as EMP_Name,
    m.emp_id as Manager_id,
    m.emp_name as Manager_name
FROM employee e
LEFT JOIN employee m 
ON e.manager_id = m.emp_id
```

```python
from pyspark.sql.functions import *

joined_df = employee_df.alias("e").join(
    employee_df.alias("m"), 
    col("e.manager_id") == col("m.emp_id"),
    "left"
).select(
    col("e.emp_id").alias("EMP_id"),
    col("e.emp_name").alias("EMP_Name"),
    col("m.emp_id").alias("Manager_id"),
    col("m.emp_name").alias("Manager_name")
)

display(joined_df)
```
